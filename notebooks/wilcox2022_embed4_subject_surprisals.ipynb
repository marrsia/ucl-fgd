## Multi-model

```python
gpt2_subject_df = pd.read_csv(gpt2_surprisals)
gpt2_object_df = pd.read_csv(object_gpt2_surprisals)

gpt2_df = pd.concat([gpt2_subject_df, gpt2_object_df], axis = 0)

grnn_subject_df = pd.read_csv(grnn_surprisals)
grnn_object_df = pd.read_csv(object_grnn_surprisals)

grnn_df = pd.concat([grnn_subject_df, grnn_object_df], axis = 0)



ngram_subject_df = pd.read_csv(ngram_surprisals)
ngram_object_df = pd.read_csv(object_ngram_surprisals)

ngram_df = pd.concat([ngram_subject_df, ngram_object_df], axis = 0)

model_dfs = {
    "GPT-2": gpt2_subject_df,
    "GRNN":  grnn_subject_df,
    "ngram": ngram_subject_df,
}

data.plot_effect_bar_chart_multi_model(
    model_dfs     = model_dfs,
    effect_fn     = data.get_filler_offloading_effect,
    conditions    = ["+wh", "-wh"],
    title         = "Filler-offloading effect by model and embedding level -- local (first token)",
    ylabel        = "Surprisal difference (bits)",
    surprisal_type= "local",
    gap_type      = "subject",
    condition_labels = ["gap", "no_gap"]
)
```

```python
data.plot_effect_bar_chart_multi_model(
    model_dfs     = model_dfs,
    effect_fn     = data.get_filler_offloading_effect,
    conditions    = ["+wh", "-wh"],
    title         = "Local (range mean) Filler-offloading by model and embedding level -- subject gap",
    ylabel        = "Surprisal difference (bits)",
    surprisal_type= "local_mean",
    gap_type      = "subject",
    condition_labels = ["+wh", "-wh"]
)
```
