## Multi-model

```python
gpt2_df = pd.read_csv(gpt2_surprisals)
grnn_df = pd.read_csv(grnn_surprisals)
ngram_df = pd.read_csv(ngram_surprisals)

model_dfs = {
    "GPT-2": gpt2_df,
    "GRNN":  grnn_df,
    "ngram": ngram_df,
}

data.plot_effect_bar_chart_multi_model(
    model_dfs     = model_dfs,
    effect_fn     = data.get_wh_effect,
    conditions    = ["gap", "no_gap"],
    title         = "Wh-effect by model and embedding level",
    ylabel        = "Surprisal difference (bits)",
    surprisal_type= "local",
    gap_type      = "subject",
    condition_labels = ["gap", "no_gap"]
)
```

```python
data.plot_effect_bar_chart_multi_model(
    model_dfs     = model_dfs,
    effect_fn     = data.get_filler_offloading_effect,
    conditions    = ["+wh", "-wh"],
    title         = "Semi Local Filler-offloading by model and embedding level",
    ylabel        = "Surprisal difference (bits)",
    surprisal_type= "semi_local",
    gap_type      = "object",
    condition_labels = ["+wh", "-wh"]
)
```
