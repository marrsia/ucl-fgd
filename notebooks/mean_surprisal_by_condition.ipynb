```python
surprisal_df = pd.read_csv(gpt2_object_csv_path)

data.plot_mean_surprisal_by_condition(
    surprisal_df,
    title='Mean Local (region mean) Surprisal by Condition and Embedding Level (GPT-2, Object Gap)',
    conditions=['+wh_gap', '+wh_no_gap', '-wh_gap', '-wh_no_gap'],
    condition_labels=['+wh gap', '+wh no-gap', '-wh gap', '-wh no-gap'],
    gap_type="object",
    surprisal_type="local_mean"
)
```

```python
surprisal_df = pd.read_csv(gpt2_object_csv_path)

data.plot_surprisal_type_difference_by_condition(
    surprisal_df,
    title='Diff between local-first-token and local-region-mean Surprisal by Condition and Embedding Level (GPT-2, Object Gap)',
    conditions=['+wh_gap', '+wh_no_gap', '-wh_gap', '-wh_no_gap'],
    condition_labels=['+wh gap', '+wh no-gap', '-wh gap', '-wh no-gap'],
    gap_type="object",
    surprisal_type_1="local",
    surprisal_type_2="local_mean"
)
## for +gap local-first-token > local-mean, for -gap local-first-token < local-mean
## continuation in particular (no gap condition) regions tend to be pretty large: later words affect surprisals more
```
